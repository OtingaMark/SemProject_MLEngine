2.10 Mapping High Blood Pressure and High Cholesterol

The CDC's 2017 calculated variables _RFHYPE5 and _RFCHOL1 indicate whether respondents have been told by a health professional that they have high blood pressure or high cholesterol.

In the source variables, the code 1 represents No and code 2 represents Yes. The existing model dataset uses the opposite binary representation: 0 for No and 1 for Yes.

Therefore, the valid source codes will be reversed when creating the corresponding model features. Unknown, refused, or missing responses will remain missing and will not be automatically classified as No.

```python
# Create a separate DataFrame for preparing 2017 model features.
# The original raw survey data remains unchanged.

df_2017_prepared = pd.DataFrame(index=df_2017.index)

# Convert the CDC high blood pressure indicator:
# source 1 = No, source 2 = Yes, source 9 = unknown/missing.
df_2017_prepared["HighBP"] = df_2017["_RFHYPE5"].map({
    1.0: 0.0,
    2.0: 1.0
})

# Convert the CDC high cholesterol indicator:
# source 1 = No, source 2 = Yes.
# Missing source values remain missing.
df_2017_prepared["HighChol"] = df_2017["_RFCHOL1"].map({
    1.0: 0.0,
    2.0: 1.0
})

# Inspect the converted values and any missing observations.
for feature in ["HighBP", "HighChol"]:
    print(f"\n--- {feature} ---")
    print(df_2017_prepared[feature].value_counts(dropna=False).sort_index())
```

2.11 Preparing the Cholesterol Check Feature

The original 2015 dataset represents CholCheck as a binary feature: 0 means the respondent has never had their cholesterol checked, while 1 means they have had it checked.

According to the CDC's 2017 BRFSS codebook, CHOLCHK1 records how long ago the respondent last had their cholesterol checked. Code 1 means never; codes 2, 3, 4, and 5 represent different time periods when a check occurred. Codes 7 and 9, as well as missing values, indicate responses that cannot be reliably classified.

To maintain consistency with the original model dataset, codes 2–5 will be mapped to 1, code 1 to 0, and uncertain or missing responses will remain missing.

```python
# Convert CHOLCHK1 into the same binary representation as the
# original 2015 CholCheck feature.
# The raw 2017 survey data remains unchanged.

cholcheck_mapping = {
    1.0: 0.0,  # Never had cholesterol checked
    2.0: 1.0,  # Checked within the past year
    3.0: 1.0,  # Checked within the past 2 years
    4.0: 1.0,  # Checked within the past 5 years
    5.0: 1.0   # Checked 5 or more years ago
}

df_2017_prepared["CholCheck"] = (
    df_2017["CHOLCHK1"].map(cholcheck_mapping)
)

# Compare the prepared feature with the original dataset's coding.
print("2015 CholCheck distribution:")
print(df_2015["CholCheck"].value_counts(dropna=False).sort_index())

print("\nPrepared 2017 CholCheck distribution:")
print(
    df_2017_prepared["CholCheck"]
    .value_counts(dropna=False)
    .sort_index()
)
```

2.13 Preparing Smoking Status

The original 2015 dataset represents Smoker as a binary feature indicating whether a respondent has smoked at least 100 cigarettes in their lifetime.

The CDC's 2017 _SMOKER3 variable distinguishes current daily smokers, current occasional smokers, former smokers, and people who have never smoked at least 100 cigarettes. To maintain consistency with the original feature, codes 1, 2, and 3 will be mapped to 1, while code 4 will be mapped to 0. Code 9 represents unknown, refused, or missing responses and will remain missing.

This mapping preserves the original feature's lifetime-smoking definition rather than treating only current smokers as smokers.

```python
# Convert the four-level CDC smoking-status variable into the
# binary representation used in the original 2015 dataset.
# The raw 2017 data remains unchanged.

smoker_mapping = {
    1.0: 1.0,  # Current smoker: every day
    2.0: 1.0,  # Current smoker: some days
    3.0: 1.0,  # Former smoker; smoked at least 100 cigarettes
    4.0: 0.0   # Never smoked at least 100 cigarettes
}

df_2017_prepared["Smoker"] = df_2017["_SMOKER3"].map(smoker_mapping)

# Compare the original and prepared feature distributions.
print("2015 Smoker distribution:")
print(df_2015["Smoker"].value_counts(dropna=False).sort_index())

print("\nPrepared 2017 Smoker distribution:")
print(
    df_2017_prepared["Smoker"]
    .value_counts(dropna=False)
    .sort_index()
)
```

```python
# Refine the two feature mappings using the 2017 source variables
# that more closely match the original feature definitions.
# The raw 2017 survey data remains unchanged.

# _MICHD: 1 = history of heart attack or coronary heart disease,
# 2 = neither condition.
df_2017_prepared["HeartDiseaseorAttack"] = (
    df_2017["_MICHD"].map({
        1.0: 1.0,
        2.0: 0.0
    })
)

# TOLDHI2: 1 = told high cholesterol, 2 = not told high cholesterol.
# Codes 7 and 9 and missing values remain missing.
df_2017_prepared["HighChol"] = (
    df_2017["TOLDHI2"].map({
        1.0: 1.0,
        2.0: 0.0
    })
)

# Inspect the refined feature distributions.
for feature in ["HeartDiseaseorAttack", "HighChol"]:
    print(f"\n--- Refined {feature} ---")
    print(
        df_2017_prepared[feature]
        .value_counts(dropna=False)
        .sort_index()
    )
```

```python

# Step: Inspect feature data types and observed ranges.
# This helps us choose an appropriate imputation strategy.

feature_columns = list(df_2015.columns.drop("Diabetes_012"))

feature_profile = pd.DataFrame({
    "Data_Type": df_2017_prepared[feature_columns].dtypes.astype(str),
    "Unique_Values": df_2017_prepared[feature_columns].nunique(),
    "Minimum": df_2017_prepared[feature_columns].min(),
    "Maximum": df_2017_prepared[feature_columns].max(),
    "Missing_Count": df_2017_prepared[feature_columns].isna().sum()
})

display(feature_profile)
```

```python

# Step: Combine the labelled 2015 and 2017 records.
# Preserve the original dataframes and keep only the shared features.

feature_columns = list(df_2015.columns.drop("Diabetes_012"))
required_columns = feature_columns + ["Diabetes_012"]

# Prepare copies with identical column order.
df_2015_combined = df_2015[required_columns].copy()
df_2015_combined["SurveyYear"] = 2015

df_2017_combined = df_2017_prepared[required_columns].copy()
df_2017_combined = df_2017_combined.dropna(subset=["Diabetes_012"]).copy()
df_2017_combined["SurveyYear"] = 2017

# Combine without changing the original datasets.
df_combined = pd.concat(
    [df_2015_combined, df_2017_combined],
    ignore_index=True
)

# Keep the target as an integer class label.
df_combined["Diabetes_012"] = (
    df_combined["Diabetes_012"].astype("int64")
)

print("Combined dataset shape:", df_combined.shape)
print("\nRecords by survey year:")
print(df_combined["SurveyYear"].value_counts().sort_index())

print("\nTarget counts by survey year:")
display(pd.crosstab(df_combined["SurveyYear"], df_combined["Diabetes_012"]))

print("\nCombined target counts:")
print(df_combined["Diabetes_012"].value_counts().sort_index())

print("\nMissing target values:", df_combined["Diabetes_012"].isna().sum())
```

```python

# Step 5: Create a stratified train/test split.
# Imputation will be fitted on the training set only.

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

# Separate predictors and target.
feature_columns = list(df_2015.columns.drop("Diabetes_012"))

X = df_combined[feature_columns].copy()
y = df_combined["Diabetes_012"].copy()

# Numerical measurements.
numeric_features = ["BMI", "MentHlth", "PhysHlth"]

# Binary and category-coded features.
categorical_features = [
    col for col in feature_columns
    if col not in numeric_features
]

# Stratified split preserves class proportions.
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Define imputation separately for the two feature groups.
preprocessor = ColumnTransformer(
    transformers=[
        (
            "numeric",
            SimpleImputer(strategy="median"),
            numeric_features
        ),
        (
            "categorical",
            SimpleImputer(strategy="most_frequent"),
            categorical_features
        )
    ],
    remainder="drop"
)

# Fit only on training data; transform both sets.
X_train_imputed = preprocessor.fit_transform(X_train)
X_test_imputed = preprocessor.transform(X_test)

print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))
print("Training class distribution:")
print(y_train.value_counts().sort_index())

print("\nTesting class distribution:")
print(y_test.value_counts().sort_index())

print("\nImputed training shape:", X_train_imputed.shape)
print("Imputed testing shape:", X_test_imputed.shape)
print("Missing values after training imputation:",
      int(__import__("numpy").isnan(X_train_imputed).sum()))
print("Missing values after test imputation:",
      int(__import__("numpy").isnan(X_test_imputed).sum()))
```

```python

# Step 6: Train a baseline Decision Tree on the combined dataset.
# Compare the result with the original 2015-only model.

from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    f1_score
)

combined_tree = DecisionTreeClassifier(random_state=42)

combined_tree.fit(X_train_imputed, y_train)

y_pred_combined = combined_tree.predict(X_test_imputed)

print("COMBINED 2015 + 2017 DECISION TREE RESULTS")
print("-" * 50)

print("Accuracy:", round(accuracy_score(y_test, y_pred_combined), 4))
print("Macro F1:", round(
    f1_score(y_test, y_pred_combined, average="macro"), 4
))

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred_combined,
    labels=[0, 1, 2],
    target_names=["No diabetes", "Prediabetes", "Diabetes"],
    digits=4,
    zero_division=0
))

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred_combined, labels=[0, 1, 2]))
```


## Experiment 1: Baseline Decision Tree Using Combined 2015 and 2017 Data

### Objective
To investigate whether combining the 2015 and 2017 BRFSS datasets improves the classification of diabetes-related categories, particularly prediabetes and diabetes.

### Methodology
The datasets were combined using the 21 shared input features and the `Diabetes_012` target. Records with missing target labels were excluded. The combined data was divided into training and testing sets using an 80:20 stratified split with `random_state=42`.

Missing numerical feature values were imputed using the training-set median, while binary and category-coded features were imputed using the training-set most frequent value. The imputation pipeline was fitted only on the training data to prevent data leakage.

The experiment will use the same training and testing sets, imputed features, and random seed as Experiment 1. Keeping these conditions constant allows us to compare the effect of class weighting more fairly.

Performance will be evaluated using accuracy, macro F1, per-class precision and recall, and a confusion matrix. Particular attention will be given to prediabetes recall and macro F1 because the dataset is imbalanced.

## Experiment 7: Stratified Cross-Validation for Model Selection

### Objective
To compare promising class-weighted Decision Tree configurations using cross-validation on the training data.

### Methodology
Five-fold stratified cross-validation will be performed using maximum tree depths of 5, 8, 10, and 20. Each model will use `class_weight="balanced"` and `random_state=42`.

The scoring metric will be macro F1, which gives equal weight to the F1 scores of the no-diabetes, prediabetes, and diabetes classes. The mean and standard deviation of macro F1 across the five folds will be reported.

The held-out test set will not be used for model selection.

### Important Implementation Detail
Imputation must be fitted separately within each cross-validation training fold to prevent data leakage. Therefore, the imputation steps and classifier will be combined into a single scikit-learn pipeline.


```python

# Experiment 7: Five-fold stratified cross-validation.
# Imputation is fitted independently within each fold.

from sklearn.model_selection import StratifiedKFold, cross_validate
from sklearn.pipeline import Pipeline
from sklearn.tree import DecisionTreeClassifier
import pandas as pd

cv_depths = [5, 8, 10, 20]

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

cv_results = []

for depth in cv_depths:
    # Recreate the imputation pipeline for this experiment.
    fold_preprocessor = ColumnTransformer(
        transformers=[
            (
                "numeric",
                SimpleImputer(strategy="median"),
                numeric_features
            ),
            (
                "categorical",
                SimpleImputer(strategy="most_frequent"),
                categorical_features
            )
        ],
        remainder="drop"
    )

    model_pipeline = Pipeline([
        ("imputer", fold_preprocessor),
        (
            "classifier",
            DecisionTreeClassifier(
                max_depth=depth,
                class_weight="balanced",
                random_state=42
            )
        )
    ])

    scores = cross_validate(
        model_pipeline,
        X_train,
        y_train,
        cv=cv,
        scoring={
            "macro_f1": "f1_macro",
            "accuracy": "accuracy"
        },
        n_jobs=-1,
        return_train_score=False
    )

    cv_results.append({
        "Max_Depth": depth,
        "Mean_CV_Macro_F1": scores["test_macro_f1"].mean(),
        "Std_CV_Macro_F1": scores["test_macro_f1"].std(),
        "Mean_CV_Accuracy": scores["test_accuracy"].mean()
    })

cv_results_df = pd.DataFrame(cv_results)

display(cv_results_df.round(4))
```


## Results and Interpretation: Stratified Cross-Validation

Five-fold stratified cross-validation was performed on the training data to compare class-weighted Decision Trees with maximum depths of 5, 8, 10, and 20.

The model with a maximum depth of 8 achieved the highest mean cross-validation macro F1 score of 0.4082, with a standard deviation of 0.0032. Its mean cross-validation accuracy was 0.6013.

The depth-20 model achieved the highest mean accuracy of 0.6788 but a lower mean macro F1 score of 0.4026. This indicates that higher accuracy did not correspond to the best balance of performance across the three target classes.

### Conclusion

Based on mean cross-validation macro F1, maximum depth 8 is the preferred candidate among the four configurations evaluated. The candidate will now be assessed on the held-out test set. The test results will provide an estimate of performance on unseen records from this combined dataset.


```python

# Step 19: Final held-out test evaluation for the selected candidate.
# Selection was based on training-set cross-validation macro F1.

from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    f1_score
)

final_candidate = Pipeline([
    ("imputer", ColumnTransformer(
        transformers=[
            (
                "numeric",
                SimpleImputer(strategy="median"),
                numeric_features
            ),
            (
                "categorical",
                SimpleImputer(strategy="most_frequent"),
                categorical_features
            )
        ],
        remainder="drop"
    )),
    ("classifier", DecisionTreeClassifier(
        max_depth=8,
        class_weight="balanced",
        random_state=42
    ))
])

final_candidate.fit(X_train, y_train)
y_pred_final = final_candidate.predict(X_test)

print("SELECTED MODEL: BALANCED DECISION TREE, MAX DEPTH 8")
print("-" * 60)
print("Accuracy:", round(accuracy_score(y_test, y_pred_final), 4))
print("Macro F1:", round(
    f1_score(y_test, y_pred_final, average="macro"), 4
))

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred_final,
    labels=[0, 1, 2],
    target_names=["No diabetes", "Prediabetes", "Diabetes"],
    digits=4,
    zero_division=0
))

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred_final, labels=[0, 1, 2]))
```

```python

# Step 42: Evaluate prediabetes thresholds using training-set cross-validation.

import numpy as np
import pandas as pd

from sklearn.base import clone
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import (
    accuracy_score,
    f1_score,
    precision_score,
    recall_score
)

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

thresholds = [0.10, 0.20, 0.30]
threshold_results = []

for fold, (train_idx, val_idx) in enumerate(
    cv.split(X_train, y_train), start=1
):
    X_fold_train = X_train.iloc[train_idx]
    X_fold_val = X_train.iloc[val_idx]
    y_fold_train = y_train.iloc[train_idx]
    y_fold_val = y_train.iloc[val_idx]

    fold_model = clone(rf_pipeline)
    fold_model.fit(X_fold_train, y_fold_train)

    probabilities = fold_model.predict_proba(X_fold_val)
    classes = list(fold_model.named_steps["classifier"].classes_)
    prediabetes_index = classes.index(1)

    original_predictions = fold_model.predict(X_fold_val)
    prediabetes_probabilities = probabilities[:, prediabetes_index]

    for threshold in thresholds:
        adjusted_predictions = original_predictions.copy()
        adjusted_predictions[prediabetes_probabilities >= threshold] = 1

        threshold_results.append({
            "Fold": fold,
            "Threshold": threshold,
            "Accuracy": accuracy_score(
                y_fold_val, adjusted_predictions
            ),
            "Macro F1": f1_score(
                y_fold_val, adjusted_predictions,
                average="macro", zero_division=0
            ),
            "Prediabetes Precision": precision_score(
                y_fold_val, adjusted_predictions,
                labels=[1], average="macro", zero_division=0
            ),
            "Prediabetes Recall": recall_score(
                y_fold_val, adjusted_predictions,
                labels=[1], average="macro", zero_division=0
            )
        })

threshold_results_df = pd.DataFrame(threshold_results)

print("MEAN 5-FOLD CROSS-VALIDATION RESULTS")
display(
    threshold_results_df.groupby("Threshold")[
        [
            "Accuracy",
            "Macro F1",
            "Prediabetes Precision",
            "Prediabetes Recall"
        ]
    ].mean().round(4)
)
```


### 7. Random Forest Probability-Threshold Evaluation

#### Objective
To investigate whether adjusting the probability threshold for the prediabetes class improves its detection rate.

#### Methodology
The balanced Random Forest was evaluated using five-fold stratified cross-validation on the training dataset. Thresholds of 0.10, 0.20, and 0.30 were tested. When a record's predicted probability for prediabetes met the selected threshold, its predicted class was changed to prediabetes.

#### Results
A threshold of 0.10 achieved an average accuracy of 0.7213, macro F1-score of 0.4316, prediabetes precision of 0.0296, and prediabetes recall of 0.1845. A threshold of 0.20 achieved the highest macro F1-score of 0.4455, while a threshold of 0.30 achieved the highest accuracy of 0.7900.

#### Interpretation
Lowering the threshold to 0.10 substantially improved prediabetes recall compared with the higher thresholds, but precision remained very low. This indicates that increasing the number of detected prediabetes cases also creates many false-positive predictions.

#### Conclusion
Threshold adjustment provides a way to investigate the trade-off between detecting more prediabetes cases and limiting false positives. The threshold of 0.20 achieved the highest macro F1-score among the three tested settings, while 0.10 offered higher prediabetes recall. Neither result alone establishes that the model is suitable for screening. Further evaluation must consider the intended use, false-positive burden, and limitations of the dataset.
